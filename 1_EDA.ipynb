```python
plot = (ggplot(df, aes(x='total_comp'))
        + geom_histogram(bins=30))

plot
```

```python
plot = (ggplot(df, aes(x="total_comp", y="assoc_year", color="region"))
        + geom_point(alpha=0.5, size=2)
        + labs(
            title="My Plot Title", x="X Axis Label", y="Y Axis Label"
            )
        +facet_grid('region')
            + theme_minimal()
            
        )

plot
```
